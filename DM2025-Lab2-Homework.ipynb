```python
import json
import re
import numpy as np
import pandas as pd

import torch
from tqdm import tqdm
from sklearn.metrics import f1_score
from torch.optim import AdamW
from torch.utils.data import Dataset, DataLoader
from sklearn.model_selection import train_test_split
from transformers import RobertaTokenizer, RobertaForSequenceClassification
```

```python
#The Different File Paths
POSTS = "/home/felix/Desktop/sentiment/data/final_posts.json"
EMOTIONS = "/home/felix/Desktop/sentiment/data/emotion.csv"
IDS = "/home/felix/Desktop/sentiment/data/data_identification.csv"
SAMPLE = "/home/felix/Desktop/sentiment/data/samplesubmission.csv"
```

```python
#  Helper Function for the json file(Formatting )
def load_posts(path: str) -> pd.DataFrame:

    with open(path, "r", encoding="utf-8") as f:
        raw_posts = json.load(f)

    posts = []
    for item in raw_posts:
        post = item["root"]["_source"]["post"]
        posts.append(
            {
                "post_id": post.get("post_id"),
                "text": post.get("text", ""),
                "hashtags": post.get("hashtags", []),
            }
        )

    df_posts = pd.DataFrame(posts)
    df_posts["hashtags_str"] = df_posts["hashtags"].apply(lambda tags: " ".join(tags or []))
    return df_posts
```

```python
# Loading the files
df_posts = load_posts(POSTS)
df_emotion = pd.read_csv(EMOTIONS)
df_id = pd.read_csv(IDS)
df_sample = pd.read_csv(SAMPLE)

print("Posts:", df_posts.shape)
print("Emotions:", df_emotion.shape)
print("ID:", df_id.shape)
print("Sample submission:", df_sample.shape)
```

```python
# naming giving consistent names 
df_id = df_id.rename(columns={"id": "post_id", "split": "data_type"})
df_emotion = df_emotion.rename(columns={"id": "post_id"})
```

```python
# putting it all in one df
df = (
    df_posts
    .merge(df_id, on="post_id", how="left")
    .merge(df_emotion, on="post_id", how="left")
)
```

```python
# looking at the data
print(df.head())
print(df["data_type"].value_counts(dropna=False))
print(df["emotion"].value_counts(dropna=False))
```

```python
# I use regex to clean the text data
URL_RE = re.compile(r"http\S+")
USER_RE = re.compile(r"@\w+")
HASHTAG_RE = re.compile(r"#(\w+)")

# Cleaing the data by lowering the case, removing urls , users, removing the hashtags(because twitter), making the whitespaces normal
def cleaning(text: str) -> str:
    text = str(text)

    text = text.lower()

    text = URL_RE.sub(" URL ", text)
    text = USER_RE.sub(" USER ", text)

    text = HASHTAG_RE.sub(r"\1", text)

    text = re.sub(r"\s+", " ", text).strip()
    return text


# combine text and hashtags for the model
def createtextinput(row):
    text = row["text"] if pd.notnull(row["text"]) else ""

    # put hashtag into a list
    hashtags = row["hashtags"] if isinstance(row["hashtags"], list) else []

    tags = " ".join(hashtags).strip()

    text = text.strip()

    # here i add [SEP] only if hashtags exist
    if len(tags) > 0:
        combined = f"{text} [SEP] {tags}"
    else:
        combined = text

    return cleaning(combined)



df["clean_text"] = df.apply(createtextinput, axis=1)

# split for training and also testing
train_df = df[df["data_type"] == "train"].copy()
test_df = df[df["data_type"] == "test"].copy()
```

```python
print("Train size:", train_df.shape)
print("Test size:", test_df.shape)
```

## 2. Feature Engineering Steps

```python
### Add the code related to the feature engineering steps in cells inside this section
```

```python
train_df[["text", "hashtags", "clean_text"]]
```

```python
# apply label encoding
labels = sorted(train_df["emotion"].unique())
label2id = {label: i for i, label in enumerate(labels)}
id2label = {i: label for label, i in label2id.items()}

train_df["label_id"] = train_df["emotion"].map(label2id)

print("Label mapping:", label2id)
```

```python
#  Train and Validation Split
train_df2, val_df = train_test_split(
    train_df,
    test_size=0.1,
    stratify=train_df["emotion"],
    random_state=42
)

print("Train size:", len(train_df2))
print("Validation size:", len(val_df))
```

```python
# apply tokenizer
tokenizer = RobertaTokenizer.from_pretrained("roberta-base")
```

```python
# Dataset class for pytorch
class EmotionDataset(Dataset):
    def __init__(self, df):
        self.texts = df["clean_text"].tolist()
        self.labels = df["label_id"].tolist()

    def __len__(self):
        return len(self.texts)

    def __getitem__(self, idx):
        enc = tokenizer(
            self.texts[idx],
            truncation=True,
            padding="max_length",
            max_length=128,
            return_tensors="pt"
        )

        return {
            "input_ids": enc["input_ids"].squeeze(),
            "attention_mask": enc["attention_mask"].squeeze(),
            "labels": torch.tensor(self.labels[idx], dtype=torch.long)
        }
```

## 3. Model Implementation Steps

```python
### Add the code related to the model implementation steps in cells inside this section
```

```python

# use data loaders
train_dataset = EmotionDataset(train_df2)
val_dataset   = EmotionDataset(val_df)

train_loader = DataLoader(train_dataset, batch_size=16, shuffle=True)
val_loader   = DataLoader(val_dataset, batch_size=32, shuffle=False)

print("-DataLoaders ready.")
```

```python
#  model
model = RobertaForSequenceClassification.from_pretrained(
    "roberta-base",
    num_labels=len(label2id)
).cuda()

optimizer = AdamW(model.parameters(), lr=2e-5)

EPOCHS = 4  
```

```python
#Training part
for epoch in range(EPOCHS):
    print(f"Epoch {epoch + 1} of {EPOCHS}")

    model.train()
    train_losses = []

    for batch in tqdm(train_loader, desc="Training"):
        batch = {k: v.cuda() for k, v in batch.items()}

        outputs = model(**batch)
        loss = outputs.loss

        loss.backward()
        optimizer.step() 
        optimizer.zero_grad()

        train_losses.append(loss.item())

    avg_train_loss = sum(train_losses) / len(train_losses)
    print(f"Train Loss: {avg_train_loss:.4f}")
```

```python
#Validation part
model.eval()
preds, truths = [], []

with torch.no_grad():
    for batch in tqdm(val_loader, desc="Validating"):
        labels = batch["labels"].cpu().numpy()
        batch = {k: v.cuda() for k, v in batch.items()}

        logits = model(**batch).logits
        pred = torch.argmax(logits, axis=1).cpu().numpy()

        preds.extend(pred)    
        truths.extend(labels) 

f1 = f1_score(truths, preds, average="macro")
print(f"Validation Macro F1: {f1:.4f}")
```

```python

model.eval()
test_preds = []
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

for text in test_df["clean_text"]:
    enc = tokenizer(
        text,
        max_length=128,
        padding="max_length",
        truncation=True,
        return_tensors="pt"
    )
    enc = {k: v.to(device) for k, v in enc.items()}

    with torch.no_grad():
        logits = model(**enc).logits

    pred_id = torch.argmax(logits, axis=1).item()
    pred_label = id2label[pred_id]
    test_preds.append(pred_label)

submission = pd.DataFrame({
    "id": test_df["post_id"],
    "emotion": test_preds
})

submission_path = "submission.csv"
submission.to_csv(submission_path, index=False)

print("Submission saved to:", submission_path)
submission.head()
```
